#**DATA AUGUMENTATION**

```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import Sequential
from keras.preprocessing import image
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
img=image.load_img('/content/cAT.jpeg', target_size=(200,200))
img
```

```python
type(img)
```

```python
img=image.img_to_array(img)
img.shape
```

```python
input_batch=img.reshape(1,200,200,3)
```

```python
data=ImageDataGenerator(
    rotation_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    width_shift_range=0.2,
    height_shift_range=0.2,
    fill_mode='nearest'
)
```

##**generat images using single image**

```python
i=0
for data in data.flow(input_batch, batch_size=1, save_to_dir='aug'):  #input_bathc=image data in array form, batch_size=1 image at once time
  i+=1
  if i==10:
    break
```

##**Generat image using bulk images from folder**

```python
train=ImageDataGenerator(
    rescale=1./255,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True
)

test=ImageDataGenerator(rescale=1./255)

train_gen=train.flow_from_directory(
    '/content/dogcat/train' ,
    target_size=(150,150) ,
    batch_size=16 ,
    class_mode='binary'
)

test_gen=test.flow_from_directory(
    '/content/dogcat/validation' ,
    target_size=(150,150) ,
    batch_size=16,
    class_mode='binary'
  )
```
